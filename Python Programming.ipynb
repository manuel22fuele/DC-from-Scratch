### Recursion

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else: return fibonacci(n-1) + fibonacci(n-2)
```

```python
fibonacci(6)
```

```python

def somaAnterior(n):
    if n <= 1:
        return n
    return n + somaAnterior(n-1) 
```

```python
somaAnterior(10)
```
