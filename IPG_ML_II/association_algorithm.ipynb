## Apriori Algorithm

```python
import numpy as np 
import pandas as pd
from apyori import apriori 
```

```python
store_dataset = pd.read_csv('./Datasets/store_data.csv')
store_dataset.head()
```

```python
store_dataset = pd.read_csv('./Datasets/store_data.csv', header=None)
```

```python
store_dataset.head()
```

#### Converting the data frame into a list

```python
all_records = []
for i in range(0, 7501):
    all_records.append([str(store_dataset.values[i, j]) for j in range(0, 20)])
```

```python
apriori_rules = apriori(all_records, min_support=0.003, min_confidence=0.1, min_lift=2, min_length=2)
apriori_rules = list(apriori_rules)
print(len(apriori_rules))
```

```python
print(apriori_rules[0])
```

```python
for rule in apriori_rules:
    item_pair = rule[0]
    items = [x for x in item_pair]
    print("The apriori rule is: " + items[0] + "->" + items[1])
    
    print("The support for the rule is: " + str(rule[1]))
    print("The confidence for the rule is: " + str(rule[2][0][2]))
    print("The lift for the rule is: " + str(rule[2][0][3]))
    print("***********************")
```

## F-P Algorithm

```python
from mlxtend.frequent_patterns import fpgrowth
from mlxtend.preprocessing import TransactionEncoder
import pandas as pd
```

```python
# Create example dataset

dataset = [['milk','bread','butter'], ['milk', 'bread'], ['milk','butter'],
           ['bread', 'butter'], ['milk','bread', 'butter', 'eggs'], ['eggs', 'butter']]
```

```python
# Create transactional encoder and transform data
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)
```

```python
# Generate frequent itemsets
frequent_itemsets = fpgrowth(df, min_support=0.05, use_colnames=True)

# Sort frequent itemsets by support
frequent_itemsets=frequent_itemsets.sort_values(by='support', ascending=False)

#Print top 10 frequent itemsets
print(frequent_itemsets.head(10))
```
