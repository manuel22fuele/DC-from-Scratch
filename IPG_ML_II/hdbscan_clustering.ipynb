```python

import numpy as np
import matplotlib.pyplot as plt
import hdbscan
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score, adjusted_rand_score

# Generate simple data

X, y_true = make_blobs(n_samples=1000, centers=5, cluster_std=0.6, random_state=42)
```

```python
# Initialize HDBSCAN with additional parameters

clusterer = hdbscan.HDBSCAN(
    min_cluster_size=5, 
    min_samples=5, 
    cluster_selection_method='eom', 
    allow_single_cluster=True, 
    metric='euclidean', 
    algorithm='best', 
    leaf_size=30
    )
```

```python
# Fit the model

clusterer.fit(X)
```

```python
# Evaluate clustering
labels = clusterer.labels_
silhouette_avg = silhouette_score(X, labels)


# Evaluate metrics
print('silhouette coefficient: {:.2f}'.format(silhouette_avg))
```

```python
# Visualizing the clusters

plt.figure(figsize=(10, 6))
plt.scatter(X[:, 0], X[:, 1], c=clusterer.labels_, cmap='viridis', s=50, alpha=0.7, edgecolors='k')
plt.colorbar()
plt.title('HDBSCAn Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```
